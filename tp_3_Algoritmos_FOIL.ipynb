#3️⃣ ** Desarrollar un programa en Python que muestre la Regla inducida para identificar a personas en formación:**

```python
# Dataset de ejemplo
datos = [ 
    {"edad": 22, "departamento": "IT", "nivel_educativo": "terciario", "en_formacion": True}, 
    {"edad": 24, "departamento": "IT", "nivel_educativo": "universitario", "en_formacion": True}, 
    {"edad": 21, "departamento": "RRHH", "nivel_educativo": "terciario", "en_formacion": True}, 
    {"edad": 35, "departamento": "IT", "nivel_educativo": "universitario", "en_formacion": False}, 
    {"edad": 40, "departamento": "Finanzas", "nivel_educativo": "maestría", "en_formacion": False}, 
    {"edad": 29, "departamento": "RRHH", "nivel_educativo": "universitario", "en_formacion": False}, 
    {"edad": 23, "departamento": "IT", "nivel_educativo": "terciario", "en_formacion": True}, 
    {"edad": 38, "departamento": "Finanzas", "nivel_educativo": "universitario", "en_formacion": False}
]

# Separar ejemplos positivos y negativos
positivos = [p for p in datos if p["en_formacion"]]
negativos = [p for p in datos if not p["en_formacion"]]

# Algoritmo FOIL simplificado: inducir condiciones exclusivas de los ejemplos positivos
def inducir_regla(positivos, negativos):
    atributos = ["edad", "departamento", "nivel_educativo"]
    regla = {}

    for atributo in atributos:
        valores_pos = set(p[atributo] for p in positivos)
        valores_neg = set(p[atributo] for p in negativos)

        if atributo == "edad":
            valores_validos = sorted([v for v in valores_pos if v not in valores_neg])
        else:
            valores_validos = list(valores_pos - valores_neg)

        if valores_validos:
            regla[atributo] = valores_validos

    return regla

# Ejecutar el algoritmo
regla_inducida = inducir_regla(positivos, negativos)

# Mostrar la regla inducida
print("Regla inducida para identificar a personas en formación:")
for atributo, valores in regla_inducida.items():
    print(f"- {atributo} debe ser uno de: {valores}")
```

>>🔹 nivel_educativo debe ser uno de: ['terciario']

# 🧠 Ejercicio 2: Cálculo de FOIL Gain

## 📌 a) Programa en Python para evaluar la condición `nivel_educativo == 'terciario'

Desarrollar un programa en Python que obtenga el cálculo del **FOIL Gain** para la condición `nivel_educativo == 'terciario'`, generando exactamente la siguiente salida:

* 🔹 **Condición:** `nivel_educativo == 'terciario'`
* 🔹 **P (positivos antes)** = `4`
* 🔹 **N (negativos antes)** = `4`
* 🔹 **p (positivos después)** = `3`
* 🔹 **n (negativos después)** = `0`
* 🔹 **p / (p + n)** = `1.000`
* 🔹 **P / (P + N)** = `0.500`
* 🔹 **log2(p / (p + n))** = `0.000`
* 🔹 **log2(P / (P + N))** = `-1.000`
* 🔹 **FOIL Gain** = `3.000`

```python
import math

# Dataset del ejercicio anterior
datos = [ 
    {"edad": 22, "departamento": "IT", "nivel_educativo": "terciario", "en_formacion": True}, 
    {"edad": 24, "departamento": "IT", "nivel_educativo": "universitario", "en_formacion": True}, 
    {"edad": 21, "departamento": "RRHH", "nivel_educativo": "terciario", "en_formacion": True}, 
    {"edad": 35, "departamento": "IT", "nivel_educativo": "universitario", "en_formacion": False}, 
    {"edad": 40, "departamento": "Finanzas", "nivel_educativo": "maestría", "en_formacion": False}, 
    {"edad": 29, "departamento": "RRHH", "nivel_educativo": "universitario", "en_formacion": False}, 
    {"edad": 23, "departamento": "IT", "nivel_educativo": "terciario", "en_formacion": True}, 
    {"edad": 38, "departamento": "Finanzas", "nivel_educativo": "universitario", "en_formacion": False}
]

# 1. Valores ANTES de aplicar la condición
P = sum(1 for d in datos if d["en_formacion"])
N = sum(1 for d in datos if not d["en_formacion"])

# 2. Filtrar aplicando la condición: nivel_educativo == 'terciario'
filtrados = [d for d in datos if d["nivel_educativo"] == "terciario"]

# 3. Valores DESPUÉS de aplicar la condición
p = sum(1 for d in filtrados if d["en_formacion"])
n = sum(1 for d in filtrados if not d["en_formacion"])

# 4. Cálculo de proporciones y logaritmos
frac_despues = p / (p + n)
frac_antes = P / (P + N)

log_despues = math.log2(frac_despues) if frac_despues > 0 else 0.0
log_antes = math.log2(frac_antes)

# 5. Cálculo del FOIL Gain
foil_gain = p * (log_despues - log_antes)

# 6. Salida formateada
print("Condición: nivel_educativo == 'terciario'")
print(f"P (positivos antes) = {P}")
print(f"N (negativos antes) = {N}")
print(f"p (positivos después) = {p}")
print(f"n (negativos después) = {n}")
print(f"p / (p + n) = {frac_despues:.3f}")
print(f"P / (P + N) = {frac_antes:.3f}")
print(f"log2(p / (p + n)) = {log_despues:.3f}")
print(f"log2(P / (P + N)) = {log_antes:.3f}")
print(f"FOIL Gain = {foil_gain:.3f}")
```

# 📊 Ejercicio 2: Cálculo de FOIL Gain

## 🧠 b) Cálculo de FOIL Gain Manual para comprobar los resultados

### 📌 Condición: `nivel_educativo == 'terciario'


### 📋 **Valores Utilizados:**

* 🟢 **P (positivos antes):** `4`
* 🔴 **N (negativos antes):** `4`
* 🟢 **p (positivos después):** `3`
* 🔴 **n (negativos después):** `0`

### 🔢 **Desarrollo Paso a Paso:**

* 1️⃣ **Fracción después de aplicar la condición:**
  $$\frac{p}{p + n} = \frac{3}{3 + 0} = \frac{3}{3} = 1.000$$

* 2️⃣ **Fracción antes de aplicar la condición:**
  $$\frac{P}{P + N} = \frac{4}{4 + 4} = \frac{4}{8} = 0.500$$

* 3️⃣ **Logaritmos en base 2:**
  * $\log_2(1.000) = 0.000$
  * $\log_2(0.500) = -1.000$

* 4️⃣ **Cálculo del FOIL Gain:**
  $$\text{FOIL Gain} = p \times \left( \log_2\left(\frac{p}{p+n}\right) - \log_2\left(\frac{P}{P+N}\right) \right)$$
  $$\text{FOIL Gain} = 3 \times (0.000 - (-1.000)) = 3 \times 1.000 = \mathbf{3.000}$$

### 💡 **5. Interpretación de los resultados**

* 🎯 **Precisión perfecta:**  
  La condición `nivel_educativo == 'terciario'` atrapa a **3 empleados en formación** ($p = 3$) y a **ningún negativo** ($n = 0$). Esto significa que no comete errores ni produce falsos positivos.

* 📈 **Ganancia de información alta (FOIL Gain = 3.000):**  
  Como la precisión dentro del grupo filtrado subió al **100%** (frente al **50%** inicial), el algoritmo obtiene la máxima ganancia posible para la cantidad de ejemplos positivos encontrados

* 📌 **Conclusión simple:**  
  Esta condición es una **excelente regla** para el modelo, ya que todo lo que clasifica como "terciario" es efectivamente una persona en formación
